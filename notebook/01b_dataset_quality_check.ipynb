```python
import os
from PIL import Image

print("Libraries loaded")
```

```python
dataset_path = "../dataset/level1_dataset"
```

```python
bad_images = []

supported_formats = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".gif",
    ".bmp",
    ".tiff"
)


for root, dirs, files in os.walk(dataset_path):

    for file in files:

        if file.lower().endswith(supported_formats):

            path = os.path.join(
                root,
                file
            )

            try:

                img = Image.open(path)

                img.load()

                img.close()


            except Exception as e:

                bad_images.append(
                    {
                        "file": path,
                        "error": str(e)
                    }
                )


print(
    "Bad Images Found:",
    len(bad_images)
)
```
