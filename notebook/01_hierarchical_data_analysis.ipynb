```python
import os
import matplotlib.pyplot as plt
import pandas as pd

print("Libraries loaded")
```

```python
dataset_path = "../dataset/raw"
```

```python
main_categories = os.listdir(dataset_path)

print(main_categories)
```

```python
def count_images(folder):

    total = 0

    for root, dirs, files in os.walk(folder):

        for file in files:

            if file.lower().endswith(
                (".jpg",".jpeg",".png",".webp",".gif")
            ):
                total += 1

    return total



category_counts = {}


for category in main_categories:

    path = os.path.join(
        dataset_path,
        category
    )

    category_counts[category] = count_images(path)



category_counts
```

```python
plt.figure(figsize=(8,5))


plt.bar(
    category_counts.keys(),
    category_counts.values()
)


plt.title(
    "Waste Main Category Distribution"
)


plt.xlabel(
    "Category"
)


plt.ylabel(
    "Number of Images"
)


plt.xticks(rotation=45)


plt.show()
```

```python
subcategory_counts = {}


for main_category in main_categories:

    main_path = os.path.join(
        dataset_path,
        main_category
    )


    for subcategory in os.listdir(main_path):

        sub_path = os.path.join(
            main_path,
            subcategory
        )


        subcategory_counts[subcategory] = count_images(
            sub_path
        )



subcategory_counts
```

```python
import os
import shutil


source = "../dataset/raw"

destination = "../dataset/level1_dataset"


categories = [
    "Hazardous",
    "Organic",
    "Recyclable",
    "Non-Recyclable"
]


os.makedirs(destination, exist_ok=True)


for category in categories:

    dest_folder = os.path.join(
        destination,
        category
    )

    os.makedirs(
        dest_folder,
        exist_ok=True
    )


    category_path = os.path.join(
        source,
        category
    )


    for subcategory in os.listdir(category_path):

        sub_path = os.path.join(
            category_path,
            subcategory
        )


        if os.path.isdir(sub_path):

            for image in os.listdir(sub_path):

                source_file = os.path.join(
                    sub_path,
                    image
                )

                destination_file = os.path.join(
                    dest_folder,
                    image
                )


                shutil.copy(
                    source_file,
                    destination_file
                )


print("Level 1 dataset created")
```
